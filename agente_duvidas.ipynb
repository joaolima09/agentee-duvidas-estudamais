## 1. Preparação do Ambiente Python

Esta célula verifica se o ambiente possui Python e confirma que o notebook está pronto.

```python
import sys

print("Versão do Python:", sys.version)
print("Ambiente pronto para executar o Agente de Dúvidas!")
```

## 2. Definição da Classe do Agente

A classe representa o agente. O método `__init__` guarda o nome em `self.nome`, enquanto `responder` processará a pergunta.

```python
class AgenteDuvidas:

    def __init__(self, nome):
        self.nome = nome

    def responder(self, pergunta):
        pergunta = pergunta.lower().strip()

        if pergunta == "o que é python?":
            return "Python é uma linguagem de programação."

        elif pergunta == "o que é api?":
            return "API é uma forma de comunicação entre sistemas."

        elif pergunta == "o que é banco de dados?":
            return "Banco de dados é um sistema que armazena informações organizadas."

        elif pergunta == "o que é programação?":
            return "Programação é o processo de criar instruções para que computadores executem tarefas."

        return "Desculpe, ainda não sei responder essa pergunta."
```

## 3. Regras Fixas e Normalização

A implementação usa `lower()` para ignorar maiúsculas e `strip()` para remover espaços das pontas. Cada pergunta cadastrada é comparada literalmente por `==`.

## 4. Entrada de Perguntas pelo Usuário

```python
agente = AgenteDuvidas("Tutor IA")

pergunta = input("Digite sua dúvida: ")
print(agente.responder(pergunta))
```

## 5. Teste Manual das Respostas

```python
casos_manuais = [
    ("O que é Python?", "Python é uma linguagem de programação."),
    ("O QUE É API?", "API é uma forma de comunicação entre sistemas."),
    ("O que é banco de dados?", "Banco de dados é um sistema que armazena informações organizadas."),
    ("  o que é API?  ", "API é uma forma de comunicação entre sistemas."),
    ("O que é python", "Desculpe, ainda não sei responder essa pergunta."),
    ("O que e Python?", "Desculpe, ainda não sei responder essa pergunta."),
]

for pergunta, esperado in casos_manuais:
    resposta = agente.responder(pergunta)
    print(f"Pergunta: {pergunta}")
    print(f"Resposta: {resposta}")
    print(f"Esperado: {esperado}")
    print("Resultado:", "OK" if resposta == esperado else "ERRO")
    print("-" * 50)
```

## 6. Testes Automatizados com Assert

```python
assert agente.nome == "Tutor IA"
assert agente.responder("O que é Python?") == "Python é uma linguagem de programação."
assert agente.responder("O QUE É API?") == "API é uma forma de comunicação entre sistemas."
assert agente.responder("  o que é api?  ") == "API é uma forma de comunicação entre sistemas."
assert agente.responder("O que é banco de dados?") == "Banco de dados é um sistema que armazena informações organizadas."
assert agente.responder("O que é programação?") == "Programação é o processo de criar instruções para que computadores executem tarefas."
assert agente.responder("Pergunta desconhecida") == "Desculpe, ainda não sei responder essa pergunta."

print("✅ Todos os testes passaram!")
```

## 7. Novas Perguntas

Duas regras foram adicionadas antes do retorno padrão: banco de dados e programação.

```python
novas_perguntas = {
    "o que é banco de dados?": "Banco de dados é um sistema que armazena informações organizadas.",
    "o que é programação?": "Programação é o processo de criar instruções para que computadores executem tarefas.",
}

for pergunta, resposta_esperada in novas_perguntas.items():
    resposta = agente.responder(pergunta)
    print(pergunta, "→", resposta)
    assert resposta == resposta_esperada

print("✅ Novas regras validadas!")
```

## 8. Execução das Perguntas em Lote

```python
casos = [
    ("O que é Python?", "Python é uma linguagem de programação."),
    ("O QUE É API?", "API é uma forma de comunicação entre sistemas."),
    ("O que é banco de dados?", "Banco de dados é um sistema que armazena informações organizadas."),
    ("O que é programação?", "Programação é o processo de criar instruções para que computadores executem tarefas."),
    ("O que é nuvem?", "Desculpe, ainda não sei responder essa pergunta."),
]

for entrada, esperado in casos:
    resultado = agente.responder(entrada)
    status = "PASSOU" if resultado == esperado else "FALHOU"
    print(f"{status} | {entrada} -> {resultado}")
```
